```python
!pip install -q \
    scikit-learn \
    xgboost \
    lightgbm \
    joblib \
    seaborn

import os
import json
import warnings
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder
)

from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)

from xgboost import XGBClassifier
from lightgbm import LGBMClassifier

warnings.filterwarnings("ignore")


RANDOM_STATE = 42


DATA_DIR = Path(
    "/content/drive/MyDrive/Razorpay_AI/ml/classification"
)

OUTPUT_DIR = Path(
    "/content/drive/MyDrive/Razorpay_AI/classification_models"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

TRAIN_PATH = DATA_DIR / "train.csv"
VAL_PATH = DATA_DIR / "validation.csv"
TEST_PATH = DATA_DIR / "test.csv"

TARGET = "primary_label"


train = pd.read_csv(TRAIN_PATH)
validation = pd.read_csv(VAL_PATH)
test = pd.read_csv(TEST_PATH)

print("=" * 80)
print("DATASET LOADED")
print("=" * 80)

print(f"Train      : {train.shape}")
print(f"Validation : {validation.shape}")
print(f"Test       : {test.shape}")


assert TARGET in train.columns, (
    f"Target '{TARGET}' missing from train.csv"
)

assert TARGET in validation.columns, (
    f"Target '{TARGET}' missing from validation.csv"
)

assert TARGET in test.columns, (
    f"Target '{TARGET}' missing from test.csv"
)

assert set(train.columns) == set(validation.columns), (
    "Train/validation columns do not match."
)

assert set(train.columns) == set(test.columns), (
    "Train/test columns do not match."
)

print("\nDataset schema: OK")


EXPECTED_CLASSES = [
    "AMOUNT_DISCREPANCY",
    "DESCRIPTION_MISMATCH",
    "DUPLICATE",
    "FORMAT_MISMATCH",
    "MISSING_RECORD",
    "TIMING_DIFFERENCE",
    "WRONG_REFERENCE"
]

observed_classes = sorted(
    train[TARGET].astype(str).unique()
)

missing_expected = sorted(
    set(EXPECTED_CLASSES) -
    set(observed_classes)
)

unexpected_classes = sorted(
    set(observed_classes) -
    set(EXPECTED_CLASSES)
)

if missing_expected:
    raise ValueError(
        f"Expected classes missing: {missing_expected}"
    )

if unexpected_classes:
    raise ValueError(
        f"Unexpected classes found: {unexpected_classes}"
    )

print("\nClass taxonomy: OK")


print("\n")
print("=" * 80)
print("CLASS DISTRIBUTION")
print("=" * 80)

for name, df in [
    ("TRAIN", train),
    ("VALIDATION", validation),
    ("TEST", test)
]:

    print(f"\n{name}")

    distribution = pd.DataFrame({

        "count":
            df[TARGET].value_counts(),

        "percentage":
            (
                df[TARGET]
                .value_counts(normalize=True)
                .mul(100)
                .round(2)
            )
    })

    display(distribution)


FORBIDDEN_COLUMNS = {

    "case_id",
    "pair_id",
    "source_id",
    "target_id",
    "affected_entity_id",

    "label",
    "primary_label",
    "fine_label",

    "ground_truth",
    "ground_truth_label",
    "expected_label",
    "expected_result",
    "true_label",
    "exception_class",

    "prediction",
    "predicted_label",
    "model_prediction",
    "match_confidence",
    "reconciliation_status",
    "final_decision",
    "human_decision"
}

DROP_COLUMNS = [
    col
    for col in train.columns
    if col in FORBIDDEN_COLUMNS
]

print("\n")
print("=" * 80)
print("EXCLUDED NON-FEATURE COLUMNS")
print("=" * 80)

print(DROP_COLUMNS)


X_train = train.drop(
    columns=DROP_COLUMNS,
    errors="ignore"
).copy()

X_val = validation.drop(
    columns=DROP_COLUMNS,
    errors="ignore"
).copy()

X_test = test.drop(
    columns=DROP_COLUMNS,
    errors="ignore"
).copy()

y_train_text = train[TARGET].astype(str)
y_val_text = validation[TARGET].astype(str)
y_test_text = test[TARGET].astype(str)


class_names = sorted(
    y_train_text.unique()
)

class_to_index = {
    label: index
    for index, label in enumerate(class_names)
}

index_to_class = {
    index: label
    for label, index in class_to_index.items()
}

y_train = y_train_text.map(
    class_to_index
).astype(int)

y_val = y_val_text.map(
    class_to_index
).astype(int)

y_test = y_test_text.map(
    class_to_index
).astype(int)

print("\n")
print("=" * 80)
print("LABEL ENCODING")
print("=" * 80)

for index, label in index_to_class.items():
    print(f"{index} -> {label}")


numeric_features = (
    X_train
    .select_dtypes(
        include=[np.number]
    )
    .columns
    .tolist()
)

categorical_features = (
    X_train
    .select_dtypes(
        include=[
            "object",
            "category",
            "bool"
        ]
    )
    .columns
    .tolist()
)

print("\n")
print("=" * 80)
print("FEATURES")
print("=" * 80)

print(
    f"Numeric features     : "
    f"{len(numeric_features)}"
)

print(
    f"Categorical features : "
    f"{len(categorical_features)}"
)

print(
    f"Total features       : "
    f"{len(X_train.columns)}"
)

print("\nNumeric:")
print(numeric_features)

print("\nCategorical:")
print(categorical_features)


missing_report = pd.DataFrame({

    "feature":
        X_train.columns,

    "missing_count":
        X_train.isna().sum().values,

    "missing_pct":
        (
            X_train.isna().mean().values
            * 100
        )
})

print("\n")
print("=" * 80)
print("MISSING VALUE REPORT")
print("=" * 80)

display(
    missing_report.sort_values(
        "missing_pct",
        ascending=False
    )
)


constant_features = [
    col
    for col in X_train.columns
    if X_train[col].nunique(
        dropna=False
    ) <= 1
]

print("\n")
print("=" * 80)
print("CONSTANT FEATURE CHECK")
print("=" * 80)

if constant_features:

    print(
        "Constant features detected:",
        constant_features
    )

else:

    print("No constant features.")


duplicate_feature_pairs = []

columns = X_train.columns.tolist()

for i in range(len(columns)):

    for j in range(i + 1, len(columns)):

        c1 = columns[i]
        c2 = columns[j]

        s1 = X_train[c1]
        s2 = X_train[c2]

        try:

            equal = s1.fillna(
                "__MISSING__"
            ).astype(str).equals(
                s2.fillna(
                    "__MISSING__"
                ).astype(str)
            )

        except Exception:

            equal = False

        if equal:

            duplicate_feature_pairs.append(
                (c1, c2)
            )

print("\n")
print("=" * 80)
print("DUPLICATE FEATURE CHECK")
print("=" * 80)

if duplicate_feature_pairs:

    for pair in duplicate_feature_pairs:
        print(pair)

else:

    print(
        "No byte-identical feature columns."
    )


class_counts = (
    y_train
    .value_counts()
    .sort_index()
)

total_samples = len(y_train)
n_classes = len(class_names)

class_weights = {}

for class_index, count in class_counts.items():

    class_weights[int(class_index)] = (
        total_samples /
        (
            n_classes *
            count
        )
    )

print("\n")
print("=" * 80)
print("CLASS WEIGHTS")
print("=" * 80)

for index, weight in class_weights.items():

    print(
        f"{index_to_class[index]:30s}"
        f"{weight:.4f}"
    )


numeric_pipeline_lr = Pipeline([

    (
        "imputer",
        SimpleImputer(
            strategy="median"
        )
    ),

    (
        "scaler",
        StandardScaler()
    )
])

categorical_pipeline_lr = Pipeline([

    (
        "imputer",
        SimpleImputer(
            strategy="most_frequent"
        )
    ),

    (
        "onehot",
        OneHotEncoder(
            handle_unknown="ignore"
        )
    )
])


preprocessor_lr = ColumnTransformer([

    (
        "numeric",
        numeric_pipeline_lr,
        numeric_features
    ),

    (
        "categorical",
        categorical_pipeline_lr,
        categorical_features
    )
])



numeric_pipeline_tree = Pipeline([

    (
        "imputer",
        SimpleImputer(
            strategy="median"
        )
    )
])

categorical_pipeline_tree = Pipeline([

    (
        "imputer",
        SimpleImputer(
            strategy="most_frequent"
        )
    ),

    (
        "onehot",
        OneHotEncoder(
            handle_unknown="ignore"
        )
    )
])


preprocessor_tree = ColumnTransformer([

    (
        "numeric",
        numeric_pipeline_tree,
        numeric_features
    ),

    (
        "categorical",
        categorical_pipeline_tree,
        categorical_features
    )
])

models = {

    "Logistic Regression":

        Pipeline([

            (
                "preprocessor",
                preprocessor_lr
            ),

            (
                "model",
                LogisticRegression(

                    max_iter=5000,

                    class_weight="balanced",

                    random_state=
                        RANDOM_STATE
                )
            )
        ]),


    "Random Forest":

        Pipeline([

            (
                "preprocessor",
                preprocessor_tree
            ),

            (
                "model",
                RandomForestClassifier(

                    n_estimators=600,

                    min_samples_leaf=2,

                    class_weight="balanced",

                    random_state=
                        RANDOM_STATE,

                    n_jobs=-1
                )
            )
        ]),


    "XGBoost":

        Pipeline([

            (
                "preprocessor",
                preprocessor_tree
            ),

            (
                "model",
                XGBClassifier(

                    n_estimators=600,

                    max_depth=6,

                    learning_rate=0.05,

                    subsample=0.85,

                    colsample_bytree=0.85,

                    objective=
                        "multi:softprob",

                    num_class=
                        n_classes,

                    eval_metric=
                        "mlogloss",

                    random_state=
                        RANDOM_STATE,

                    n_jobs=-1
                )
            )
        ]),


    "LightGBM":

        Pipeline([

            (
                "preprocessor",
                preprocessor_tree
            ),

            (
                "model",
                LGBMClassifier(

                    n_estimators=600,

                    learning_rate=0.05,

                    num_leaves=31,

                    objective=
                        "multiclass",

                    num_class=
                        n_classes,

                    class_weight=
                        "balanced",

                    random_state=
                        RANDOM_STATE,

                    n_jobs=-1,

                    verbosity=-1
                )
            )
        ])
}


def calculate_metrics(
    y_true,
    y_pred,
    y_probability
):

    metrics = {

        "accuracy":
            accuracy_score(
                y_true,
                y_pred
            ),

        "balanced_accuracy":
            balanced_accuracy_score(
                y_true,
                y_pred
            ),

        "macro_precision":
            precision_score(
                y_true,
                y_pred,
                average="macro",
                zero_division=0
            ),

        "macro_recall":
            recall_score(
                y_true,
                y_pred,
                average="macro",
                zero_division=0
            ),

        "macro_f1":
            f1_score(
                y_true,
                y_pred,
                average="macro",
                zero_division=0
            ),

        "weighted_f1":
            f1_score(
                y_true,
                y_pred,
                average="weighted",
                zero_division=0
            )
    }

    try:

        metrics["roc_auc_ovr_macro"] = (
            roc_auc_score(
                y_true,
                y_probability,
                multi_class="ovr",
                average="macro"
            )
        )

    except Exception:

        metrics["roc_auc_ovr_macro"] = np.nan

    return metrics


trained_models = {}
validation_metrics = {}

print("\n")
print("=" * 80)
print("TRAINING FOUR CLASSIFICATION MODELS")
print("=" * 80)


for model_name, model in models.items():

    print("\n")
    print("-" * 80)
    print(f"MODEL: {model_name}")
    print("-" * 80)

    model.fit(
        X_train,
        y_train
    )

    validation_probability = (
        model.predict_proba(X_val)
    )

    validation_prediction = (
        np.argmax(
            validation_probability,
            axis=1
        )
    )

    metrics = calculate_metrics(
        y_val,
        validation_prediction,
        validation_probability
    )

    trained_models[
        model_name
    ] = model

    validation_metrics[
        model_name
    ] = metrics

    print(
        f"Accuracy           : "
        f"{metrics['accuracy']:.4f}"
    )

    print(
        f"Balanced Accuracy  : "
        f"{metrics['balanced_accuracy']:.4f}"
    )

    print(
        f"Macro Precision    : "
        f"{metrics['macro_precision']:.4f}"
    )

    print(
        f"Macro Recall       : "
        f"{metrics['macro_recall']:.4f}"
    )

    print(
        f"Macro F1           : "
        f"{metrics['macro_f1']:.4f}"
    )

    print(
        f"Weighted F1        : "
        f"{metrics['weighted_f1']:.4f}"
    )

    print(
        f"ROC-AUC OVR Macro  : "
        f"{metrics['roc_auc_ovr_macro']:.4f}"
    )


comparison_rows = []

for model_name, metrics in validation_metrics.items():

    comparison_rows.append({

        "model":
            model_name,

        **metrics
    })


comparison_df = pd.DataFrame(
    comparison_rows
)

comparison_df = (
    comparison_df
    .sort_values(
        [
            "macro_f1",
            "macro_recall",
            "balanced_accuracy"
        ],
        ascending=False
    )
    .reset_index(drop=True)
)

print("\n")
print("=" * 80)
print("MODEL COMPARISON — VALIDATION")
print("=" * 80)

display(comparison_df)


best_model_name = (
    comparison_df.iloc[0]["model"]
)

best_model = trained_models[
    best_model_name
]

best_validation_metrics = (
    validation_metrics[
        best_model_name
    ]
)

print("\n")
print("=" * 80)
print("BEST MODEL")
print("=" * 80)

print(
    f"Model              : "
    f"{best_model_name}"
)

print(
    f"Validation Macro F1 : "
    f"{best_validation_metrics['macro_f1']:.4f}"
)

print(
    f"Validation Macro Recall : "
    f"{best_validation_metrics['macro_recall']:.4f}"
)


test_probability = (
    best_model
    .predict_proba(X_test)
)

test_prediction = (
    np.argmax(
        test_probability,
        axis=1
    )
)

test_metrics = calculate_metrics(
    y_test,
    test_prediction,
    test_probability
)

print("\n")
print("=" * 80)
print("FINAL TEST RESULTS")
print("=" * 80)

for metric_name, value in test_metrics.items():

    if pd.isna(value):

        print(
            f"{metric_name:25s}: N/A"
        )

    else:

        print(
            f"{metric_name:25s}: "
            f"{value:.4f}"
        )


cm = confusion_matrix(
    y_test,
    test_prediction,
    labels=list(
        range(n_classes)
    )
)

cm_df = pd.DataFrame(

    cm,

    index=[
        index_to_class[i]
        for i in range(n_classes)
    ],

    columns=[
        index_to_class[i]
        for i in range(n_classes)
    ]
)

print("\n")
print("=" * 80)
print("CONFUSION MATRIX")
print("=" * 80)

display(cm_df)


plt.figure(
    figsize=(11, 9)
)

sns.heatmap(
    cm_df,
    annot=True,
    fmt="d",
    cmap="viridis",
    cbar=True
)

plt.title(
    f"Confusion Matrix — {best_model_name}"
)

plt.xlabel(
    "Predicted Class"
)

plt.ylabel(
    "Actual Class"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.yticks(
    rotation=0
)

plt.tight_layout()

plt.show()


test_report = classification_report(

    y_test,

    test_prediction,

    labels=list(
        range(n_classes)
    ),

    target_names=[
        index_to_class[i]
        for i in range(n_classes)
    ],

    output_dict=True,

    zero_division=0
)

test_report_df = (
    pd.DataFrame(
        test_report
    )
    .transpose()
)

print("\n")
print("=" * 80)
print("PER-CLASS TEST PERFORMANCE")
print("=" * 80)

display(test_report_df)


error_rows = []

for class_index in range(n_classes):

    actual_mask = (
        y_test ==
        class_index
    )

    support = int(
        actual_mask.sum()
    )

    correct = int(
        (
            actual_mask
            &
            (
                test_prediction ==
                class_index
            )
        ).sum()
    )

    incorrect = (
        support -
        correct
    )

    error_rows.append({

        "class":
            index_to_class[
                class_index
            ],

        "support":
            support,

        "correct":
            correct,

        "incorrect":
            incorrect,

        "recall":
            correct /
            max(support, 1)
    })


error_analysis_df = pd.DataFrame(
    error_rows
)

print("\n")
print("=" * 80)
print("CLASS ERROR ANALYSIS")
print("=" * 80)

display(
    error_analysis_df
)


confusion_rows = []

for actual in range(n_classes):

    for predicted in range(n_classes):

        if actual == predicted:
            continue

        count = int(
            cm[actual, predicted]
        )

        if count > 0:

            confusion_rows.append({

                "actual":
                    index_to_class[
                        actual
                    ],

                "predicted":
                    index_to_class[
                        predicted
                    ],

                "count":
                    count
            })


top_confusions_df = pd.DataFrame(
    confusion_rows
)

print("\n")
print("=" * 80)
print("NON-ZERO CONFUSIONS")
print("=" * 80)

if top_confusions_df.empty:

    print(
        "No misclassification pairs."
    )

else:

    top_confusions_df = (
        top_confusions_df
        .sort_values(
            "count",
            ascending=False
        )
        .reset_index(drop=True)
    )

    display(
        top_confusions_df
    )


test_confidence = (
    test_probability.max(
        axis=1
    )
)

confidence_summary = pd.Series(
    test_confidence
).describe()

print("\n")
print("=" * 80)
print("TEST CONFIDENCE")
print("=" * 80)

display(
    confidence_summary
    .to_frame(
        "confidence"
    )
)


predicted_labels = [
    index_to_class[
        int(index)
    ]
    for index in test_prediction
]

confidence_df = pd.DataFrame({

    "predicted_class":
        predicted_labels,

    "confidence":
        test_confidence,

    "actual_class":
        [
            index_to_class[
                int(index)
            ]
            for index in y_test
        ]
})

low_confidence_cases = (
    confidence_df
    .sort_values(
        "confidence"
    )
    .head(50)
)

print("\n")
print("=" * 80)
print("LOW-CONFIDENCE TEST CASES")
print("=" * 80)

display(
    low_confidence_cases
)


sorted_probabilities = np.sort(
    test_probability,
    axis=1
)

top_probability = (
    sorted_probabilities[:, -1]
)

second_probability = (
    sorted_probabilities[:, -2]
)

probability_margin = (
    top_probability -
    second_probability
)

probability_analysis_df = pd.DataFrame({

    "predicted_class":
        predicted_labels,

    "top_probability":
        top_probability,

    "second_probability":
        second_probability,

    "probability_margin":
        probability_margin,

    "actual_class":
        confidence_df[
            "actual_class"
        ]
})

print("\n")
print("=" * 80)
print("PREDICTION PROBABILITY ANALYSIS")
print("=" * 80)

display(
    probability_analysis_df.head(20)
)


feature_schema = {

    "task":
        "exception_classification",

    "target":
        TARGET,

    "classes":
        class_names,

    "class_to_index":
        class_to_index,

    "index_to_class":
        {
            str(k): v
            for k, v in index_to_class.items()
        },

    "features":
        X_train.columns.tolist(),

    "numeric_features":
        numeric_features,

    "categorical_features":
        categorical_features,

    "excluded_columns":
        DROP_COLUMNS,

    "constant_features_detected":
        constant_features,

    "duplicate_feature_pairs":
        [
            list(pair)
            for pair in duplicate_feature_pairs
        ],

    "best_model":
        best_model_name
}


model_path = (
    OUTPUT_DIR /
    "best_exception_classifier.joblib"
)

joblib.dump(
    best_model,
    model_path
)


class_mapping_path = (
    OUTPUT_DIR /
    "class_mapping.json"
)

with open(
    class_mapping_path,
    "w"
) as f:

    json.dump(
        {
            "class_names":
                class_names,

            "class_to_index":
                class_to_index,

            "index_to_class":
                {
                    str(k): v
                    for k, v
                    in index_to_class.items()
                }
        },
        f,
        indent=2
    )


feature_schema_path = (
    OUTPUT_DIR /
    "feature_schema.json"
)

with open(
    feature_schema_path,
    "w"
) as f:

    json.dump(
        feature_schema,
        f,
        indent=2
    )


# ============================================================
# 35. SAVE MODEL COMPARISON
# ============================================================

comparison_path = (
    OUTPUT_DIR /
    "model_comparison.csv"
)

comparison_df.to_csv(
    comparison_path,
    index=False
)


final_metrics = {

    "task":
        "exception_classification",

    "target":
        TARGET,

    "number_of_classes":
        n_classes,

    "classes":
        class_names,

    "best_model":
        best_model_name,

    "validation_metrics":
        best_validation_metrics,

    "test_metrics":
        test_metrics,

    "dataset_sizes": {

        "train":
            int(len(train)),

        "validation":
            int(len(validation)),

        "test":
            int(len(test))
    },

    "feature_count":
        int(len(X_train.columns))
}

final_metrics_path = (
    OUTPUT_DIR /
    "final_metrics.json"
)

with open(
    final_metrics_path,
    "w"
) as f:

    json.dump(
        final_metrics,
        f,
        indent=2
    )


classification_report_path = (
    OUTPUT_DIR /
    "classification_report.csv"
)

test_report_df.to_csv(
    classification_report_path
)


confusion_matrix_path = (
    OUTPUT_DIR /
    "confusion_matrix.csv"
)

cm_df.to_csv(
    confusion_matrix_path
)


error_analysis_path = (
    OUTPUT_DIR /
    "class_error_analysis.csv"
)

error_analysis_df.to_csv(
    error_analysis_path,
    index=False
)


confidence_path = (
    OUTPUT_DIR /
    "test_prediction_confidence.csv"
)

probability_analysis_df.to_csv(
    confidence_path,
    index=False
)


model_info = {

    "best_model":
        best_model_name,

    "selection_metric":
        "validation_macro_f1",

    "random_state":
        RANDOM_STATE,

    "train_rows":
        int(len(train)),

    "validation_rows":
        int(len(validation)),

    "test_rows":
        int(len(test)),

    "features":
        X_train.columns.tolist()
}

model_info_path = (
    OUTPUT_DIR /
    "model_info.json"
)

with open(
    model_info_path,
    "w"
) as f:

    json.dump(
        model_info,
        f,
        indent=2
    )


print("\n")
print("=" * 80)
print("FINAL EXCEPTION CLASSIFICATION PIPELINE")
print("=" * 80)

print(
    f"Best model                 : "
    f"{best_model_name}"
)

print(
    f"Validation Macro F1        : "
    f"{best_validation_metrics['macro_f1']:.4f}"
)

print(
    f"Test Macro F1              : "
    f"{test_metrics['macro_f1']:.4f}"
)

print(
    f"Test Weighted F1           : "
    f"{test_metrics['weighted_f1']:.4f}"
)

print(
    f"Test Balanced Accuracy     : "
    f"{test_metrics['balanced_accuracy']:.4f}"
)

print(
    f"Test Accuracy              : "
    f"{test_metrics['accuracy']:.4f}"
)

print(
    f"Number of classes          : "
    f"{n_classes}"
)

print(
    f"Number of features         : "
    f"{len(X_train.columns)}"
)

print("\nArtifacts saved to:")
print(OUTPUT_DIR)

print("\nFiles:")

for file in sorted(
    OUTPUT_DIR.iterdir()
):

    print(
        " -",
        file.name
    )

print("\n")
print("=" * 80)
print("STATUS: READY_FOR_BACKEND_INTEGRATION")
print("=" * 80)
```
